## A. Reproducible Normalization Problem

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
```

```python
print (X)
```

```python
mean = float (X.mean())
print ("Mean: ", mean)
```

```python
std = float (X.std())
print ("Standard Deviation: ", std)
```

```python
X_normalized = (X - X.mean() ) / X.std()
```

```python
print (X_normalized)
```

```python
print ("Normalized Mean: ", X_normalized.mean())
```

```python
print ("Normalized Standard Deviation: ", X_normalized.std())
```

```python
np.save("X_normalized.npy", X_normalized)
```

## B. Cubes Divisible by 4 Problems

```python
import numpy as np
```

```python
integers = np.arange (1,101)
integers
```

```python
cubes = integers ** 3
C = cubes.reshape (10, 10)
print ("Shape of C:", C.shape)
```

```python
div_by_4 = C[C % 4 == 0]

print ("Selected array div by 4:\n", div_by_4)
```

```python
np.save("div_by_4.npy",div_by_4)
```

## C. Above-Mean Squares Problem

```python
import numpy as np
```

```python
S = (np.arange(1,37) ** 2).reshape(6,6)
print (S)
```

```python
S_mean = float (S.mean())
print ("S_mean:", S_mean)
```

```python
above_mean = S[S > S_mean]
print ("Above mean:", above_mean)
```

```python
print ("Number of selected elements:", len(above_mean))
```

```python
np.save("above_mean.npy", above_mean)
```
